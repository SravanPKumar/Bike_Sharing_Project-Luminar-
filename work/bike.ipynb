```python
df.drop(['instant','dteday','casual','registered'], axis=1, inplace=True)
```

```python
plt.figure(figsize=(8,5))
sns.lineplot(x="hr", y="cnt", data=df)
plt.title("Hourly Bike Demand Pattern")
plt.xlabel("Hour of Day")
plt.xticks(range(1, 25))
plt.ylabel("Bike Count")
plt.show()
```

```python
x = df.drop('cnt', axis=1)
y = df['cnt']
```

```python
x.shape
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.30, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# LINEAR REGRESSION
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train, y_train)
lr_pred=lr.predict(x_test)
lr_pred
```

```python
lr_mae = mean_absolute_error(y_test, lr_pred)
lr_rmse = mean_squared_error(y_test, lr_pred)
lr_r2 = r2_score(y_test, lr_pred)

print("Decision Tree Regressor")
print("MAE:", lr_mae)
print("RMSE:", lr_rmse)
print("R2 Score:", lr_r2)
```

```python
# DECISION TREE
from sklearn.tree import DecisionTreeRegressor
dt=DecisionTreeRegressor(criterion='squared_error')
dt.fit(x_train,y_train)
dt_pred=dt.predict(x_test)
dt_pred
```

```python
dt_mae = mean_absolute_error(y_test, dt_pred)
dt_rmse = mean_squared_error(y_test, dt_pred)
dt_r2 = r2_score(y_test, dt_pred)

print("Decision Tree Regressor")
print("MAE:", dt_mae)
print("RMSE:", dt_rmse)
print("R2 Score:", dt_r2)
```

```python
# RANDOM FOREST
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(n_estimators=200,random_state=42)
rf.fit(x_train, y_train)
rf_pred = rf.predict(x_test)
rf_pred
```

```python
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = mean_squared_error(y_test, rf_pred)
rf_r2 = r2_score(y_test, rf_pred)

print("Decision Tree Regressor")
print("MAE:", rf_mae)
print("RMSE:", rf_rmse)
print("R2 Score:", rf_r2)
```

```python
results = pd.DataFrame({
    "Model": ["Linear Regression", "Decision Tree Regression", "Random Forest Regression"],
    "MAE": [lr_mae, dt_mae, rf_mae],
    "RMSE": [lr_rmse, dt_rmse, rf_rmse],
    "R2 Score": [lr_r2, dt_r2, rf_r2]
})

results
```

```python
results.sort_values(by="R2 Score", ascending=False)
```

```python
pickle.dump(lr, open("Linear_Regression_Model.pkl", "wb"))
pickle.dump(dt, open("Decision_Tree_Model.pkl", "wb"))
pickle.dump(rf, open("Random_Forest_Model.pkl", "wb"))
pickle.dump(x.columns.tolist(), open("features.pkl", "wb"))

print("All models and features saved successfully")
```
